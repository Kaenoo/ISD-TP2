**3.4) À partir de la liste des pays uniques et en utilisant un outil d'IA (GPT, Copilot, etc.), créez un code qui génère, pour chaque pays, une abréviation composée des trois premières consonnes de son nom, puis affichez le résultat sous la forme `Pays -> Abréviation`. Trouvez un moyen de compléter les abréviations des pays ayant moins de trois consonnes et de gérer les éventuels doublons. Dans tous les cas, les abréviations doivent être uniques et comporter trois caractères.**

Affichez la preuve que vos abréviations sont uniques.


(3 points)


Exemple: 

```

["France", "Nicaragua", "Poland", ...]
["Frn", "Ncr", "Pln", ...]


```


**-> Indiquez dans la cellule après votre code, quel outil vous avez utilisé et quel(s) prompt a(ont) été donné(s).**


````python
# Partie 3.4 - code
import re

# Récupération des pays uniques depuis le DataFrame
countries = df['country'].unique()

abbr_dict = {}
seen_abbr = set()

for country in countries:
    # 1. Extraire uniquement les lettres (ignorer espaces, tirets, etc.) et mettre en majuscules
    letters = re.sub(r'[^a-zA-Z]', '', country).upper()
    
    # 2. Séparer les consonnes et les voyelles
    consonants = re.sub(r'[AEIOUY]', '', letters)
    vowels = re.sub(r'[^AEIOUY]', '', letters)
    
    # 3. Prendre les 3 premières consonnes
    abbr = consonants[:3]
    
    # 4. Compléter si moins de 3 consonnes (avec les voyelles du pays, puis avec des X si nécessaire)
    if len(abbr) < 3:
        abbr = (abbr + vowels)[:3]
    if len(abbr) < 3:
        abbr = (abbr + "XXX")[:3]
        
    # 5. Gestion des doublons : remplacer le(s) dernier(s) caractère(s) par un compteur
    original_abbr = abbr
    counter = 1
    while abbr in seen_abbr:
        suffix = str(counter)
        abbr = original_abbr[:3 - len(suffix)] + suffix
        counter += 1
        
    # Enregistrer l'abréviation
    seen_abbr.add(abbr)
    abbr_dict[country] = abbr

# Affichage des abréviations
for country, abbr in abbr_dict.items():
    print(f"{country} -> {abbr}")

# Preuve d'unicité
print("\n" + "-"*30)
print("PREUVE D'UNICITÉ")
print("-"*30)

all_abbr = list(abbr_dict.values())
unique_abbr = set(all_abbr) # set() élimine automatiquement les doublons

print(f"Nombre de pays dans la liste            : {len(countries)}")
print(f"Nombre d'abréviations distinctes créées : {len(unique_abbr)}")
print(f"Les abréviations sont-elles toutes uniques et de 3 lettres ? : {len(all_abbr) == len(unique_abbr) and all(len(a) == 3 for a in all_abbr)}")
````

**4.1) Calculez la moyenne de l’espérance de vie de tous les pays en 1952 et en 2007. Générez un bar chart (deux barres, une pour 1952 et une pour 2007) permettant la comparaison de ces moyennes.**



(5 points)

Pour sélectionner les données des bonnes années, regardez les [exemples de la doc ("Getting values")](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.loc.html) pour vous aider à comprendre comment faire à l'aide de conditions. 


Une fois que vous avez un nouveau DataFrame avec uniquement les données de 1952 et 2007, gardez uniquement les colonnes d'intérêt (year et lifeExp), groupez les données par année et calculez la moyenne de l'espérance de vie. Une fois ceci fait, utilisez la fonction ```df.plot.bar()```. 

Aide : vous pouvez utiliser plusieurs .loc[] d'affilée, une fois pour choisir les années et une fois pour choisir les colonnes d'intérêt. **Regardez l'exemple ci-dessous pour vous aider.**

Notes: toutes ces opérations peuvent être effectuées à la chaîne, en une seule ligne. Si vous souhaitez faire cela petit à petit, n'oubliez pas d'enregistrer vos résultats intermédiaires dans une variable: ``` tmp_df = df.loc["exemple"]```. 

**Attention : votre graphique doit contenir des labels pour les axes et un titre afin d'être lisible !**

**Pour déboguer, vous pouvez utiliser la fonction print() pour afficher le contenu intermédiaire de votre DataFrame, par exemple après avoir filtré les années. Merci de supprimer vos print() de débogage avant le rendu !**



````python
# Écrivez votre code ici

# Filtre les années 1952 et 2007 ET sélectionne les deux colonnes year et lifeExp
df_filtered = df.loc[df['year'].isin([1952, 2007]), ['year', 'lifeExp']]

# Groupe par année et calcule la moyenne de l'espérance de vie
df_mean = df_filtered.groupby('year').mean()

# Affichage intermédiare
print(df_mean)

# Trace le barchart à partir des moyennes calculées
df_mean.plot.bar(legend=False, color='skyblue')

# Ajoute le titre et les labels obligatoires
plt.title("Moyenne de l'espérance de vie mondiale : 1952 vs 2007")
plt.xlabel("Année")
plt.ylabel("Espérance de vie (en années)")

# Remet les textes de l'axe X à l'horizontale
plt.xticks(rotation=0)

# Affichage du résultat final
plt.show()

````

Ci-dessous, un petit **exemple** qui utilise le "method chaining". Cette notation, qui profite du fait que chaque méthode Pandas retourne un DataFrame auquel on peut appliquer une nouvelle méthode, augmente grandement la lisibilité du code. Vous trouverez un exemple dans [la cheat sheet "Data wrangling with Pandas"](https://pandas.pydata.org/Pandas_Cheat_Sheet.pdf). 

>Most pandas methods return a DataFrame so that another pandas method can be applied to the result. This improves readability of code.


En plus de chaîner les méthodes, le graphique est rendu beaucoup plus lisible en y ajoutant des axes bien nommés, un titre et des labels inclinés. Libre à vous de vous inspirer de cet exemple pour les exercices suivants.

````python
# Exemple

ax = (df
      .loc[:, ['continent', 'lifeExp']]
      .groupby('continent')
      .mean()
      .plot.bar(rot=45, figsize=(16, 8))
      )
ax.set_title('Overall mean of life expectancy per continent', fontsize=14)
ax.set_xlabel('Continent', fontsize=12)
ax.set_ylabel('Life expectancy', fontsize=12);
````

**4.7) Lisez attentivement le code du graphique suivant et ses commentaires. Modifiez _une seule ligne_ pour définir comme taille des points du scatterplot la population de chaque pays en millions.** Observez bien le graphique avant et après.

[Table des matières](#toc)

Vous pouvez trouver le paramètre à ajouter dans [la doc](https://pandas.pydata.org/pandas-docs/version/0.25.0/reference/api/pandas.DataFrame.plot.scatter.html). 

````python
import matplotlib.cm as cm
NCOLORS = 5

fig, axs = plt.subplots(1, 2, sharey=True, sharex=True, figsize=(20, 8))

# Colormap for the continents.
cmap = list(plt.get_cmap("rainbow")(np.linspace(0, 1, NCOLORS)))

# Vous pouvez aussi définir les couleurs manuellement
# cmap = ["black","red","yellow","blue","green"]


for i, year in enumerate((1952, 2007)):
    data = df.query('year == @year')
    
    # Plot each continent as a different color in order to have a legend. 
    
    # Observez bien: 
    # - la fonction enumerate et ce qu'elle retourne
    #     La fonction enumerate() renvoie un index en plus des éléments d'une liste. 
    #     Ici elle retourne c, l'index, ainsi qu'un tuple avec le groupe et le DataFrame correspondant.
    #     Pour accéder à la population de chaque pays on peut par exemple utiliser df_continent['pop']
    
    for c, (continent, df_continent) in enumerate(data.groupby('continent')):
        
        # Scatterplot
        df_continent.plot.scatter(ax=axs[i], 
                                  x='gdpPercap', 
                                  y='lifeExp', 
                                  color=[cmap[c]], 
                                  label=continent, 
                                  s = df_continent['pop'] / 1000000,
                                  alpha=0.8)
    
    # Plot style
    
    axs[i].set_title(year, fontsize=14)
    axs[i].set_xlabel('GDP per capita', fontsize=12)
    axs[i].grid(axis='y', linestyle='--', alpha=0.4, lw=0.5)
    
    
    # Remove border on the top and right.
    axs[i].spines['top'].set_visible(False)
    axs[i].spines['right'].set_visible(False)
    # Set alpha on remaining borders.
    axs[i].spines['left'].set_alpha(0.4)
    axs[i].spines['bottom'].set_alpha(0.4)

    # Only show ticks on the left and bottom spines.
    axs[i].xaxis.set_ticks_position('bottom')
    axs[i].yaxis.set_ticks_position('left')
    # Style of ticks.
    plt.xticks(fontsize=10, alpha=0.7)
    plt.yticks(fontsize=10, alpha=0.7)
    
axs[0].set_ylabel('Life Expectancy', fontsize=12)
fig.suptitle('Scatter plot with life expectancy by gdpPercap by country for 1952 and 2007', fontsize=16)
````
